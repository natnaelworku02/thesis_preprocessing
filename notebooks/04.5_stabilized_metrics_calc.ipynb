```python
import pandas as pd
import numpy as np
import os
import csv

# Paths (matching your workspace setup)
PROCESSED_DIR = r"C:\Users\HP\Desktop\thesis_preprocessing\data\processed"

FILES = [
    "01_Original_12k_V2_scored.csv",
    "02_Distractor_A_Swapped_12k_V2_scored.csv",
    "03_Distractor_B_Shuffled_12k_V2_scored.csv"
]

def calculate_stabilized_metrics(df):
    # 1. Ensure physical lines of code (LOC) is calculated dynamically
    df['loc'] = df['original_code'].apply(lambda x: max(len(str(x).splitlines()), 1))
    
    # Candidate 1: Log-Stabilized Adak (LSA)
    df['adak_lsa'] = np.log1p(df['mcv']) - np.log1p(df['sfv'])
    
    # Candidate 2: Linear Additive Damped Adak
    df['adak_damped'] = ((100 * df['mcv']) / (df['sfv'] + 1.0)) - 100
    
    # Candidate 3: Square-Root Damped Adak
    df['adak_sqrt'] = ((100 * df['mcv']) / np.sqrt(df['sfv'] + 1.0)) - 100
    
    # Candidate 4: Self-Stabilizing Syntactic Density Adak (adak_ss)
    df['adak_ss'] = ((100 * df['mcv'] * df['sfv']) / (df['sfv']**2 + df['loc'])) - 100
    
    return df

print("Starting calculation of stabilized metrics...")

for file_name in FILES:
    file_path = os.path.join(PROCESSED_DIR, file_name)
    if not os.path.exists(file_path):
        print(f"Error: Could not find {file_name} in {PROCESSED_DIR}")
        continue
        
    print(f"Processing: {file_name}")
    df_temp = pd.read_csv(file_path)
    
    # Calculate new columns
    df_temp = calculate_stabilized_metrics(df_temp)
    
    # Save back with proper quoting to preserve code formatting
    df_temp.to_csv(file_path, index=False, quoting=csv.QUOTE_ALL)
    print(f"✓ Saved updated {file_name}")

print("\nSUCCESS: All three scored datasets updated with LSA, Damped, Sqrt, and adak_ss columns!")
```
